```python
sql = "select * from dev_builder_ingestion.meta.etl_spark_stm where landing_table_name like '%HANDRAISERS%' \
and landing_column_name <>'' order by raw_column_order"
```

```python
df = spark.sql(sql)
display(df) 
```

```python
df2 = df[["landing_column_name" ,"landing_data_type"]]
```

```python
df2.display()
```

```python
from pyspark.sql.functions import concat, col,lit
df3 = df2.withColumn("combined", concat(col('landing_column_name'), lit(" "),col('landing_data_type')))
```

```python
df3.display()
```

```python
result = ', '.join([str(row['combined']) for row in df3.select("combined").collect()])
```
